```python
import pandas as pd

df = pd.read_csv("/kaggle/input/datasets/jessemostipak/hotel-booking-demand/hotel_bookings.csv")

# 1. Missing values
df["country"] = df["country"].fillna("Unknown")
df["children"] = df["children"].fillna(0).astype(int)

df["agent"] = df["agent"].astype("Int64")
df["company"] = df["company"].astype("Int64")

# 2. Remove impossible bookings with zero guests
df = df[
    (df["adults"] > 0) |
    (df["children"] > 0) |
    (df["babies"] > 0)
].copy()

# 3. Remove negative ADR
df = df[df["adr"] >= 0].copy()

# 4. Create technical row ID
df = df.reset_index(drop=True)
df["booking_row_id"] = df.index + 1

# 5. Create arrival date
month_map = {
    "January": 1,
    "February": 2,
    "March": 3,
    "April": 4,
    "May": 5,
    "June": 6,
    "July": 7,
    "August": 8,
    "September": 9,
    "October": 10,
    "November": 11,
    "December": 12
}

df["arrival_month_num"] = df["arrival_date_month"].map(month_map)

df["arrival_date"] = pd.to_datetime(
    dict(
        year=df["arrival_date_year"],
        month=df["arrival_month_num"],
        day=df["arrival_date_day_of_month"]
    )
)

# 6. Derived features
df["total_guests"] = (
    df["adults"] +
    df["children"] +
    df["babies"]
)

df["total_nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

df["booking_status"] = df["is_canceled"].map({
    0: "Not Cancelled",
    1: "Cancelled"
})

# 7. Estimated revenue
df["estimated_room_value"] = (
    df["adr"] * df["total_nights"]
)

df["estimated_realized_room_revenue"] = (
    df["estimated_room_value"]
    .where(df["is_canceled"] == 0, 0)
)
```
